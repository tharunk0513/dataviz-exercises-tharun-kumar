## Task 2 — A line chart showing CO2 emissions over time for all Asian countries in the dataset, with one country highlighted.

**same requirements as Task1**

```python
asia = df[df['Region'] == 'Asia'].copy()
print(asia['Country'].unique())

# First vs last year for each country
check = asia[asia['Year'].isin([2000, 2022])].pivot(index='Country', columns='Year', values='CO2_Mt')
check['Change %'] = ((check[2022] / check[2000] - 1) * 100).round(0)
print(check.round(0))
```

```python
highlight = 'China'   # change to any country from Step 1

fig = go.Figure()

# Grey context lines first, so the highlight is drawn on top
for country in asia['Country'].unique():
    if country == highlight:
        continue
    d = asia[asia['Country'] == country]
    fig.add_trace(go.Scatter(
        x=d['Year'], y=d['CO2_Mt'], mode='lines',
        line=dict(color='#CCCCCC', width=1.5),
        hoverinfo='text', hovertext=country, showlegend=False))

# Highlighted line
h = asia[asia['Country'] == highlight].sort_values('Year')
fig.add_trace(go.Scatter(
    x=h['Year'], y=h['CO2_Mt'], mode='lines',
    line=dict(color='#C0392B', width=4), showlegend=False))

# Direct label at the end of the highlighted line
fig.add_annotation(
    x=h['Year'].iloc[-1], y=h['CO2_Mt'].iloc[-1],
    text=f"<b>{highlight}</b>", showarrow=False,
    xanchor='left', xshift=8, font=dict(color='#C0392B', size=14))

fig.update_layout(
    title='YOUR INSIGHT TITLE HERE',
    plot_bgcolor='white',
    yaxis=dict(range=[0, None], title='CO₂ emissions (Mt)',
               showgrid=False, showline=False),
    xaxis=dict(title=None, showgrid=False),
    margin=dict(r=100),
    width=900, height=500,
    font=dict(size=13))
fig.show()
```
