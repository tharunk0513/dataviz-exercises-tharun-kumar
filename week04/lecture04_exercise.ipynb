## Task 2 — Bubble chart: tell a story

**What to build:** A bubble chart (full 2007 dataset, all countries) where:
- x = GDP per capita (log scale)
- y = life expectancy
- size = population
- colour = ONE continent highlighted (your choice), all others grey
- At least one annotation explaining the highlighted group's story

> This is the grey-and-highlight technique applied to a bubble chart.


```python
highlight = 'Asia'   # Africa, Americas, Asia, Europe or Oceania

d07 = df[df['year'] == 2007].copy()
sub = d07[d07['continent'] == highlight]

print(sub[['gdpPercap', 'lifeExp']].describe().round(1))
print(sub.nlargest(3, 'pop')[['country', 'pop', 'gdpPercap', 'lifeExp']])
print("Other continents, mean life expectancy:",
      d07[d07['continent'] != highlight]['lifeExp'].mean().round(1))
```

```python
import plotly.express as px

d07['group'] = d07['continent'].where(d07['continent'] == highlight, 'Other')
colour_map = {highlight: '#C0392B', 'Other': '#CCCCCC'}

# Draw grey first, so the highlighted bubbles sit on top
d07 = d07.sort_values('group', key=lambda s: s == highlight)

fig = px.scatter(
    d07, x='gdpPercap', y='lifeExp', size='pop',
    color='group', color_discrete_map=colour_map,
    hover_name='country', size_max=60, log_x=True,
    labels={'gdpPercap': 'GDP per capita (log scale, US$)',
            'lifeExp': 'Life expectancy (years)'})

fig.update_traces(marker=dict(opacity=0.75, line=dict(width=0.5, color='white')))

fig.add_annotation(
    x=3.1, y=47, xref='x', yref='y',   # x is in log10 units
    text='YOUR ANNOTATION HERE',
    showarrow=True, arrowhead=2, ax=60, ay=-40,
    font=dict(color='#C0392B', size=13))

fig.update_layout(
    title='YOUR INSIGHT TITLE HERE',
    plot_bgcolor='white', showlegend=False,
    xaxis=dict(showgrid=False),
    yaxis=dict(showgrid=False, range=[35, 87]),
    width=900, height=550, font=dict(size=13))
fig.show()
```
