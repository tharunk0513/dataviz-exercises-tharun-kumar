**Task 3 (stretch):** Build a **grouped bar chart** comparing 2 sub-factors (e.g. `GDP_per_capita` and `Freedom`) across the 5 most populated regions. Use colour meaningfully and write an insight title.

Regions to include: `'Western Europe'`, `'Latin America'`, `'East Asia'`, `'Sub-Saharan Africa'`, `'South Asia'`


```python
regions = ['Western Europe', 'Latin America and Caribbean', 'East Asia',
           'Sub-Saharan Africa', 'South Asia']

reg_avg = (df[df['Region'].isin(regions)]
           .groupby('Region')[['Social_Support', 'Freedom']]
           .mean()
           .reset_index()
           .sort_values('Social_Support', ascending=False))

print(reg_avg.round(2))
```

```python
fig = go.Figure()

fig.add_bar(x=reg_avg['Region'], y=reg_avg['Social_Support'],
            name='Social support', marker_color='#1f4e79',
            text=reg_avg['Social_Support'].round(2), textposition='outside')

fig.add_bar(x=reg_avg['Region'], y=reg_avg['Freedom'],
            name='Freedom', marker_color='#e07b00',
            text=reg_avg['Freedom'].round(2), textposition='outside')

fig.update_layout(
    barmode='group',
    title='YOUR INSIGHT TITLE HERE',
    yaxis=dict(range=[0, 1.1], title='Average score (0–1)'),
    xaxis_title=None,
    plot_bgcolor='white',
    font=dict(size=13),
    legend=dict(orientation='h', y=1.08, x=0),
    width=900, height=500,
)
fig.update_yaxes(gridcolor='#eeeeee')
fig.show()
```
